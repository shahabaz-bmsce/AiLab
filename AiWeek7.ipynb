```python
def is_variable(x):
    return isinstance(x, str) and x[0].islower()

def apply_subst(x, s):
    if isinstance(x, str):
        while x in s:
            x = s[x]
        return x
    return tuple(apply_subst(a, s) for a in x)

def occurs(v, x, s):
    x = apply_subst(x, s)
    if v == x:
        return True
    if isinstance(x, tuple):
        return any(occurs(v, a, s) for a in x)
    return False

def unify(x, y, s=None):
    if s is None:
        s = {}

    x, y = apply_subst(x, s), apply_subst(y, s)

    if x == y:
        return s

    if isinstance(x, str) and is_variable(x):
        if occurs(x, y, s):
            return None
        s[x] = y
        return s

    if isinstance(y, str) and is_variable(y):
        if occurs(y, x, s):
            return None
        s[y] = x
        return s

    if not isinstance(x, tuple) or not isinstance(y, tuple):
        return None

    if x[0] != y[0] or len(x) != len(y):
        return None

    for a, b in zip(x[1:], y[1:]):
        s = unify(a, b, s)
        if s is None:
            return None

    return s

# 1. Successful Unification
E1 = ("Eats", "x", "Apple")
E2 = ("Eats", "Riya", "y")

print("--- SUCCESSFUL UNIFICATION TEST ---")
print("Expression 1:", E1)
print("Expression 2:", E2)

result1 = unify(E1, E2, {})

print("\nRESULT:")
if result1 is not None:
    print("Status: Successful")
    print("Substitution:", result1)
    print("Expression 1:", apply_subst(E1, result1))
    print("Expression 2:", apply_subst(E2, result1))
else:
    print("Status: Failed")

print("\n" + "="*40 + "\n")

# 2. Failed Unification
E3 = ("Eats", "x", "Apple")
E4 = ("Eats", "Riya", "Mango")

print("--- FAILED UNIFICATION TEST ---")
print("Expression 1:", E3)
print("Expression 2:", E4)

result2 = unify(E3, E4, {})

print("\nRESULT:")
if result2 is not None:
    print("Status: Successful")
    print("Substitution:", result2)
else:
    print("Status: Failed")
```
