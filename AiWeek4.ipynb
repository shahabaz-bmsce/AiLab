```python

def calculate_cost(board):
    cost = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):
            if abs(board[i] - board[j]) == abs(i - j):
                cost += 1

    return cost


def print_board(board):
    n = len(board)

    for row in range(n - 1, -1, -1):
        for col in range(n):
            if board[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


def hill_climbing():
    board = [3, 1, 2, 0]

    while True:
        current_cost = calculate_cost(board)

        print("\nCurrent state:", board)
        print("Current cost:", current_cost)
        print_board(board)

        if current_cost == 0:
            print("\nGoal state reached!")
            return

        best_board = None
        best_cost = current_cost
        best_pair = None

        n = len(board)

        for i in range(n):
            for j in range(i + 1, n):
                new_board = board.copy()
                new_board[i], new_board[j] = new_board[j], new_board[i]

                cost = calculate_cost(new_board)

                print("Swap", (i, j),
                      "Neighbor:", new_board,
                      "Cost:", cost)

                if cost < best_cost:
                    best_cost = cost
                    best_board = new_board
                    best_pair = (i, j)

        if best_board is None:
            print("\nHill climbing terminated at a local minimum.")
            return

        print("Selected swap:", best_pair)
        board = best_board


hill_climbing()
```
