```python

from itertools import product

print("P\tQ\tR\tQ -> P\tP -> NOT Q\tQ OR R\tKB")

for P, Q, R in product([False, True], repeat=3):

    a = (not Q) or P
    b = (not P) or (not Q)
    c = Q or R

    KB = a and b and c

    print(
        ("T" if P else "F"), "\t",
        ("T" if Q else "F"), "\t",
        ("T" if R else "F"), "\t",
        ("T" if a else "F"), "\t",
        ("T" if b else "F"), "\t\t",
        ("T" if c else "F"), "\t",
        ("T" if KB else "F")
    )

print("\nChecking Entailment:")

def entails(query):
    for P, Q, R in product([False, True], repeat=3):

        a = (not Q) or P
        b = (not P) or (not Q)
        c = Q or R

        KB = a and b and c

        if KB and not query(P, Q, R):
            return False

    return True

print("KB entails R:", entails(lambda P, Q, R: R))
print("KB entails R -> P:", entails(lambda P, Q, R: (not R) or P))
print("KB entails Q -> R:", entails(lambda P, Q, R: (not Q) or R))
```
